```python
import pandas as pd
```

```python
# Sales data
sales = pd.DataFrame({
    "order_id": [101, 102, 103, 104, 105, 106],
    "region": ["North", "South", "East", "West", "North", "South"],
    "product": ["Laptop", "Phone", "Laptop", "Tablet", "Phone", "Tablet"],
    "sales": [55000, 30000, 60000, 25000, 35000, 28000]
})

# Customer data
customers = pd.DataFrame({
    "order_id": [101, 102, 103, 107],
    "customer": ["Anita", "Rahul", "Priya", "Kiran"],
    "city": ["Delhi", "Bangalore", "Mumbai", "Chennai"]
})

print("Sales Data:")
display(sales)

print("\nCustomer Data:")
display(customers)
```

```python
# INNER JOIN
inner_join = pd.merge(
    sales,
    customers,
    on="order_id",
    how="inner"
)

display(inner_join)
```

```python
# LEFT JOIN
left_join = pd.merge(
    sales,
    customers,
    on="order_id",
    how="left"
)

display(left_join)
```

```python
# RIGHT JOIN
right_join = pd.merge(
    sales,
    customers,
    on="order_id",
    how="right"
)

display(right_join)
```

```python
# OUTER JOIN
outer_join = pd.merge(
    sales,
    customers,
    on="order_id",
    how="outer"
)

display(outer_join)
```

```python
# Pivot table: total sales by region and product
sales_pivot = pd.pivot_table(
    sales,
    values="sales",
    index="region",
    columns="product",
    aggfunc="sum",
    fill_value=0
)

display(sales_pivot)
```

```python
# Pivot table with row and column totals
sales_pivot_totals = pd.pivot_table(
    sales,
    values="sales",
    index="region",
    columns="product",
    aggfunc="sum",
    fill_value=0,
    margins=True,
    margins_name="Total"
)

display(sales_pivot_totals)
```

```python
# Create wide-format sales data
wide_sales = pd.DataFrame({
    "region": ["North", "South", "East"],
    "Laptop": [55000, 0, 60000],
    "Phone": [35000, 30000, 0],
    "Tablet": [0, 28000, 25000]
})

display(wide_sales)
```

```python
# Convert wide format to long format
long_sales = pd.melt(
    wide_sales,
    id_vars=["region"],
    var_name="product",
    value_name="sales"
)

display(long_sales)
```

```python
# Reshape long data back into a wide format
reshaped_sales = long_sales.pivot(
    index="region",
    columns="product",
    values="sales"
).fillna(0)

display(reshaped_sales)
```

```python
# Method chaining: filter, group, calculate, and sort
chained_analysis = (
    sales
    .query("sales > 25000")
    .groupby("region", as_index=False)["sales"]
    .sum()
    .sort_values("sales", ascending=False)
)

display(chained_analysis)
```

## Conclusion

In this notebook, I practiced advanced Pandas operations including:

- `pd.merge()` for combining DataFrames using different join types.
- `pd.pivot_table()` for summarising sales by region and product.
- `pd.melt()` for converting wide-format data into long-format data.
- `pivot()` for reshaping long data back into wide format.
- Method chaining for performing multiple Pandas operations in a readable workflow.

These operations are useful for combining, reshaping, summarising, and analysing real-world business data.
