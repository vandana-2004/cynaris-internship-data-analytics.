```python
import numpy as np
import time

print("NumPy version:", np.__version__)
```

```python
# 1D array
arr_1d = np.array([10, 20, 30, 40, 50])

# 2D array
arr_2d = np.array([
    [10, 20, 30],
    [40, 50, 60]
])

# 3D array
arr_3d = np.array([
    [[1, 2], [3, 4]],
    [[5, 6], [7, 8]]
])

print("1D array:", arr_1d)
print("1D shape:", arr_1d.shape)

print("\n2D array:\n", arr_2d)
print("2D shape:", arr_2d.shape)

print("\n3D array:\n", arr_3d)
print("3D shape:", arr_3d.shape)
```

```python
sales = np.array([1000, 2000, 3000, 4000, 5000])

print("Add 500:", sales + 500)
print("Multiply by 2:", sales * 2)
print("Total sales:", np.sum(sales))
print("Mean sales:", np.mean(sales))

monthly_sales = np.array([
    [1000, 1500, 2000],
    [2500, 3000, 3500],
    [4000, 4500, 5000]
])

print("Column totals:", np.sum(monthly_sales, axis=0))
print("Row totals:", np.sum(monthly_sales, axis=1))
```

```python
sales = np.array([12000, 18000, 25000, 32000, 15000])

# Boolean masking
high_sales = sales[sales > 20000]
print("Sales above 20000:", high_sales)

# Fancy indexing
selected_sales = sales[[0, 2, 4]]
print("Selected sales:", selected_sales)
```

```python
sales = np.array([12000, 18000, 25000, 32000, 15000])

print("Mean:", np.mean(sales))
print("Standard deviation:", np.std(sales))
print("75th percentile:", np.percentile(sales, 75))

quantity = np.array([1, 2, 3, 4, 5])

correlation = np.corrcoef(sales, quantity)[0, 1]
print("Correlation:", correlation)
```

```python
n = 1_000_000
numbers = list(range(n))
numpy_numbers = np.arange(n)

# Python loop
start = time.perf_counter()

python_result = [x * 2 for x in numbers]

python_time = time.perf_counter() - start

# NumPy vectorisation
start = time.perf_counter()

numpy_result = numpy_numbers * 2

numpy_time = time.perf_counter() - start

print(f"Python loop time: {python_time:.6f} seconds")
print(f"NumPy operation time: {numpy_time:.6f} seconds")

if numpy_time > 0:
    print(f"Speed ratio: {python_time / numpy_time:.2f}x")

print("Results match:", np.array_equal(
    np.array(python_result), numpy_result
))
```

```python
prices = np.array([100, 200, 300])
discount = 10

discounted_prices = prices * (1 - discount / 100)

print("Original prices:", prices)
print("Discounted prices:", discounted_prices)
```
