```python

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# Sample business sales dataset
df = pd.DataFrame({
    "group": ["Online"] * 10 + ["Store"] * 10,
    "sales": [
        120, 150, 130, 170, 160, 140, 180, 155, 145, 500,
        110, 125, 135, 140, 150, 145, 130, 155, 160, 165
    ]
})

# Descriptive statistics
print("Mean:", df["sales"].mean())
print("Median:", df["sales"].median())
print("Mode:", df["sales"].mode().tolist())
print("Variance:", df["sales"].var())
print("Standard deviation:", df["sales"].std())

# Percentiles and quartiles
print("\n25th percentile (Q1):", df["sales"].quantile(0.25))
print("50th percentile (Q2):", df["sales"].quantile(0.50))
print("75th percentile (Q3):", df["sales"].quantile(0.75))
print("90th percentile:", df["sales"].quantile(0.90))
print("\nFive-number summary:")
print(df["sales"].describe())

# Frequency distribution
plt.figure(figsize=(8, 4))
plt.hist(df["sales"], bins=8, edgecolor="black")
plt.title("Sales Frequency Distribution")
plt.xlabel("Sales")
plt.ylabel("Frequency")
plt.tight_layout()
plt.show()

# Compare groups using side-by-side box plots
plt.figure(figsize=(7, 4))
df.boxplot(column="sales", by="group")
plt.title("Sales Distribution: Online vs Store")
plt.suptitle("")
plt.xlabel("Sales Channel")
plt.ylabel("Sales")
plt.tight_layout()
plt.show()

# Interpretation
print("\nInterpretation:")
print("The mean is sensitive to unusually high or low values.")
print("The median is more robust when the data contains outliers.")
print("Standard deviation measures the spread around the mean.")
print("A normal distribution is approximately symmetric and bell-shaped.")
print("The histogram helps identify skewness and multiple peaks.")
```
