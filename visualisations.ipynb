```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

print("Libraries loaded successfully!")

# Load the Iris dataset



iris = load_iris()

df = pd.DataFrame(
    iris.data,
    columns=iris.feature_names
)

df["species"] = pd.Categorical.from_codes(
    iris.target,
    iris.target_names
)

df.head()
```

```python
import sqlite3
import pandas as pd

# Connect to the existing sales database
conn = sqlite3.connect("sales.db")

# Check the available tables
tables = pd.read_sql_query(
    "SELECT name FROM sqlite_master WHERE type='table';",
    conn
)

tables

df = pd.read_sql_query("SELECT * FROM orders", conn)

df.head()

df.info()
```

```python
import matplotlib.pyplot as plt

# Count orders by date
orders_by_date = df.groupby("order_date").size()

# Create line chart
plt.figure(figsize=(10, 5))
plt.plot(orders_by_date.index, orders_by_date.values, marker="o")

plt.title("Number of Orders Over Time")
plt.xlabel("Order Date")
plt.ylabel("Number of Orders")
plt.xticks(rotation=45)

plt.tight_layout()
plt.show()

# Count orders by region
orders_by_region = df.groupby("region").size()

# Create bar chart
plt.figure(figsize=(8, 5))
plt.bar(orders_by_region.index, orders_by_region.values)

plt.title("Number of Orders by Region")
plt.xlabel("Region")
plt.ylabel("Number of Orders")

plt.tight_layout()
plt.show()

```

```python
print(df.columns.tolist())
```

```python
# Scatter plot: Quantity vs Unit Price
plt.figure(figsize=(8, 5))

plt.scatter(df["quantity"], df["unit_price"], alpha=0.7)

plt.title("Quantity vs Unit Price")
plt.xlabel("Quantity")
plt.ylabel("Unit Price")

plt.tight_layout()
plt.show()
```

```python
# Histogram: Distribution of Unit Prices
plt.figure(figsize=(8, 5))

plt.hist(df["unit_price"], bins=10)

plt.title("Distribution of Unit Prices")
plt.xlabel("Unit Price")
plt.ylabel("Frequency")

plt.tight_layout()
plt.show()
```

```python
# Box plot: Distribution of Unit Prices
plt.figure(figsize=(8, 5))

plt.boxplot(df["unit_price"])

plt.title("Box Plot of Unit Prices")
plt.ylabel("Unit Price")

plt.tight_layout()
plt.show()
```

```python
import seaborn as sns

# Apply Seaborn theme
sns.set_theme(style="whitegrid")

# Example chart using the Seaborn theme
plt.figure(figsize=(8, 5))

plt.bar(orders_by_region.index, orders_by_region.values)

plt.title("Number of Orders by Region")
plt.xlabel("Region")
plt.ylabel("Number of Orders")

plt.tight_layout()
plt.show()
```

```python
# Find the region with the highest number of orders
top_region = orders_by_region.idxmax()
top_orders = orders_by_region.max()

# Create the bar chart
plt.figure(figsize=(8, 5))

plt.bar(orders_by_region.index, orders_by_region.values)

plt.title("Number of Orders by Region")
plt.xlabel("Region")
plt.ylabel("Number of Orders")

# Add a data callout
plt.annotate(
    f"Highest: {top_region} ({top_orders} orders)",
    xy=(top_region, top_orders),
    xytext=(0, 20),
    textcoords="offset points",
    ha="center",
    arrowprops=dict(arrowstyle="->")
)

plt.tight_layout()
plt.show()
```
